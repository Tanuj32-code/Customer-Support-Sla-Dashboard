```python
# Library creation
import pandas as pd
import numpy as np
import random
from datetime import datetime,timedelta
```

```python
# Parameter creation
num_tickets = 500
queues = ["Billing","Technical Support","Returns", "Prime Support"]
sla_targets = [24,48] # Hours
```

```python
# Generate synthetic data
data = []
start_date = datetime(2026, 8, 1, 8, 0, 0)
```

```python
for i in range(num_tickets):
    ticket_id = f"TCKT{i+1:04d}"
    queue = random.choice(queues)
    created = start_date + timedelta(hours=random.randint(0, 240))
    resolution_time = random.randint(1, 72)  # hours to resolve
    resolved = created + timedelta(hours=resolution_time)
    sla_target = random.choice(sla_targets)
    status = "Closed" if resolved < datetime.now() else "Open"

    data.append([ticket_id, queue, created, resolved, sla_target, status])
```

```python
# Create DataFrame
df = pd.DataFrame(data, columns=[
    "Ticket_ID", "Queue_Name", "Created_DateTime",
    "Resolved_DateTime", "SLA_Target_Hours", "Status"
])
```

```python
# Add derived metrics
df["Resolution_Time_Hours"] = (df["Resolved_DateTime"] - df["Created_DateTime"]).dt.total_seconds() / 3600
df["SLA_Compliant"] = df["Resolution_Time_Hours"] <= df["SLA_Target_Hours"]
```

```python
# Save to CSV
df.to_csv("synthetic_support_tickets.csv", index=False)
```
